# 🔒 PHASE 4: Sicherheits-Architektur & Sandbox Hardening
**Projekt: Mai_AI (MaiOmni) — Datensouveränität & Sandboxing**

Als hochperformante und private KI-Plattform steht Datensicherheit an oberster Stelle. Ein lokaler KI-Agent darf niemals in der Lage sein, unkontrollierte Lese-/Schreibrechte auf dem Hostsystem zu besitzen. 

--- 
### 🎯 Sicherheits-Prinzipien dieser Phase:
1. **Ordner-Sandboxing:** Strikt getrennte Mounts für jeden Benutzer (`/app/users/{username}/`).
2. **Path Traversal Schutz:** Verhindern des Ausbrechens aus Benutzerordnern (`../../etc/passwd`).
3. **Ressourcen-Limits:** Begrenzung von Arbeitsspeicher (max. 1GB) und CPU (max. 1 Core) pro Container, um Denial-of-Service-Angriffe (DoS) zu blockieren.
4. **Sicherheits-Wächter (FileGuard):** Schutz kritischer Systemdateien.

---

### 🛡️ Schritt 1: Sandbox Pfad-Traversal Test
Wir prüfen logisch, ob Pfade korrekt normalisiert und isoliert werden, damit kein Benutzer die Daten eines anderen Benutzers einsehen kann.

```python
def is_safe_path(base_dir, path, follow_symlinks=True):
    if follow_symlinks:
        matchpath = os.path.realpath(path)
    else:
        matchpath = os.path.abspath(path)
    return base_dir == os.path.commonpath((base_dir, matchpath))

# Test-Szenario
user_home = os.path.abspath("../users/testuser")
attack_path = os.path.join(user_home, "../../etc/passwd")
safe_path = os.path.join(user_home, "workspace/my_file.txt")

print(f"Hauptordner: {user_home}")
print(f"Normaler Pfad safe? {is_safe_path(user_home, safe_path)} (Erwartet: True)")
print(f"Angriffs-Pfad (Traversal) safe? {is_safe_path(user_home, attack_path)} (Erwartet: False)")
```
