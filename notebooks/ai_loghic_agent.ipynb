## 🎯 Mission Statement & Projekt-Fokus
Dieses Jupyter Notebook (`ai_loghic_agent.ipynb`) dient als **vollständige, lückenlose Implementierungs- und Architektur-Planung** für einen hochgradig modularen, 100% lokal und offline betreibbaren KI-Agenten. 

Das Gesamtsystem löst eine fundamentale Schwachstelle klassischer lokaler LLM-Systeme: **Vage, unpräzise oder kontextarme Nutzeranfragen führen bei kleineren lokalen Modellen oft zu Halluzinationen oder unbrauchbarem Code.** 

Durch die vorausschauende Kopplung zweier spezialisierter Instanzen – dem **kognitiven Logik- und Spezifikations-Agenten (JEV / LAYA)** und dem **ausführenden Worker-Modell (`llama_cpp`)** – wird jede Benutzeranfrage vor der eigentlichen Abarbeitung kognitiv analysiert, mit dem vorhandenen Wissen der SQLite-Datenbank (`knowledge_agent_routing_tree.db`) angereichert, gegen das System-Manifest validiert und zu einem **hochpräzisen, fehlerresistenten Ziel-Prompt** veredelt.

Das visuelle Frontend wird über ein hochentwickeltes **3-Spalten-Layout in `ipywidgets` (angelehnt an die Google Antigravity IDE / Virtual Studio Optic)** realisiert:
1. **Linke Spalte (Aufklappbar & mit Breiten-Merker):** Verwaltung vergangener Themen-Anfragen (`Recent Chats`), Sessions und Routing-Knoten direkt aus der SQLite-Historie.
2. **Mittlere Spalte (Interaktive Chat-Konsole):** Dynamischer Nachrichtenverlauf mit transparentem, aufklappbarem Denkprozess (*"Show thinking"* mit Live-Inspektion der Agentenschritte) und moderner Multimodal-Eingabeleiste mit Audio- und Werkzeug-Schnittstellen.
3. **Rechte Spalte (Aufklappbar & mit Breiten-Merker - *Virtual Studio Optic*):** Vollständiges Ausführungs-Terminal, dynamische Session-Logs aus `log_YYYYMMDD_HHMMSS`, Sandbox-Audits und Fehler-Tracing (`meta_admin_repair_registry` & `FX_CHAIN`).

> **Wichtiger Hinweis zum Dokumentationsstatus:**  
> Dieses Notebook enthält gemäß Spezifikation **ausschließlich detaillierte Markdown-Planungsschritte**. Es wird kein Rohcode zur Ausführung vorgegriffen, sondern jeder einzelne Mikroschritt – von den Modul-Imports über die Audio-Verarbeitung, das interaktive 3-Spalten-Dashboard, die SQL-Abfragelogik und die Prompt-Spezifikation bis hin zur autonomen Persistenz und Tool-Planung in der Wissensdatenbank – mathematisch, algorithmisch und strukturell exakt vorgegeben.

### 0.2 Rollenverteilung im Multi-Agenten-System: JEV/LAYA vs. llama_cpp Worker

Ein zentrales architektonisches Konzept dieses Entwurfs ist die strikte Trennung von **kognitiver Spezifikation (Prompt Engineering & Routing)** und **technischer Ausarbeitung (Inferenz & Codegenerierung)**:

| Merkmal | 🧠 JEV / LAYA (Spezifikations- & Logik-Agent) | ⚙️ llama_cpp (Ausführender Task-Worker) |
| :--- | :--- | :--- |
| **Primäre Aufgabe** | Analyse der Nutzeranfrage, Eliminierung von Unschärfen, Wissensextraktion aus SQL & Projekt, Prompt-Veredelung. | Vollständige Ausarbeitung der Lösung, Generierung von Python/SQL-Code, mathematische Herleitungen, Dokumentation. |
| **Modell-Typ** | Kompaktes, hoch-instruiertes SLM/LLM (z. B. Qwen2.5-Coder-7B, Mistral-7B, Llama-3.2-3B, spezialisiertes JEV-GGUF). | Leistungsfähiges Arbeitsmodell (z. B. Codestral-22B, Qwen2.5-Coder-14B/32B, Llama-3.1-8B/70B GGUF). |
| **Temperatur-Profil** | Streng deterministisch: `temperature = 0.0 - 0.1` (Fokus auf Logik, Strukturtreue, SQL-Fakten). | Moderat deterministisch bis kreativ: `temperature = 0.2 - 0.4` für präzisen Code mit flexibler Problemlösung. |
| **Eingabe (Input)** | Rohe Benutzeranfrage (Text/Audio-Transkript) + gefilterte SQL-Tabellenzeilen + Projekt-Metadaten. | Vom JEV/LAYA veredeltes "High-Specification Augmented Prompt" (HAP) mit strikter Blueprint-Struktur. |
| **Ausgabe (Output)** | Strukturierte Zielerklärung, Akzeptanzkriterien, Pflicht-Tabellen, Taboo-Regeln, Ausführungsplan. | Lauffähiger Programmcode, Erläuterungen, Tool-Definitionen, Testfälle. |
| **Visualisierung in UI** | Erscheint im aufklappbaren **"Show thinking"**-Bereich der zentralen Chat-Konsole mit Live-Schritten. | Erscheint als formatierter Markdown-Antwortblock mit Syntax-Highlighting im Hauptverlauf. |
| **Datenbank-Interaktion** | Lesend: Abfrage von `library_registry`, `standard_column_library_registry`, `manifest`, `taboo`. | Keine direkte DB-Verbindung; wird durch Verifizierungs- und Lern-Wrapper überwacht. |

### 1.1 Umfassende Modul- und Bibliotheks-Spezifikation
Für den stabilen Betrieb aller Teilsysteme werden spezialisierte Bibliotheken benötigt. Die Planung sieht vor, die Abhängigkeiten in vier klar abgegrenzte Schichten zu gliedern:

#### 1. Standard-Bibliotheken (Python Built-ins)
* `os`, `sys`, `pathlib.Path`: Dateisystem-Traversierung, Pfad-Auflösung relativ zum Projekt-Root (`Offline_AI`), Windows-spezifische Pfadbereinigung.
* `sqlite3`: Transaktionssichere Verbindung zu `knowledge_agent_routing_tree.db` mit WAL-Modus (`Write-Ahead Logging`) und `sqlite3.Row` Row-Factory.
* `datetime`, `time`: Hochpräzise Zeitstempel-Erzeugung nach dem ISO- und dem Projektstandard (`YYYYMMDD_HHMMSS` sowie Vektoren für Jahr, Monat, Tag, Stunde, Minute, Sekunde).
* `json`, `re`: Parsing von System-Manifest-Einträgen, Token-Schärfung mittels regulärer Ausdrücke, Extraktion von Markdown-Codeblöcken.

* `subprocess`: Isolierte Prozessausführung für Sandbox-Validierung und System-Audits.

#### 2. LLM- & Inferenz-Bibliotheken
* `llama_cpp` (`from llama_cpp import Llama`): Lokale Inferenz von GGUF-quantisierten Modellen mit voller Hardware-Beschleunigung (CUDA/ROCm/Metal/OpenBLAS), Context-Window-Management und Token-Streaming.

#### 3. Audio-Schnittstellen (STT & TTS)

````python
# IMPORTE

import importlib
import subprocess
import sys

# CHAPTER 1: IMPORT ARCHITECTURE & SYSTEM REQUIREMENTS
# Extended Tuple Structure: (Import Name, Alias, Pip Install Name)++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++
LIBRARY_IMPORT = [
    # Standard-Bibliotheken & System-Utilities
    ("os",),
    ("sys",),
    ("json",),
    ("re",),
    ("time",),
    ("hashlib",),
    ("sqlite3",),
    ("threading",),
    ("queue",),
    ("ast",),
    ("subprocess",),
    ("logging",),
    ("pathlib",),
    ("datetime",),
    ("typing",),
    ("dataclasses",),
    ("shutil",),
    ("glob",),
    ("psutil",),

    # Data Science, Statistik & Visualisierung
    ("numpy", "np"),
    ("pandas", "pd"),
    ("scipy",),
    ("seaborn", "sns"),
    ("plotly",),
    ("graphviz",),

    # Netz & Web-Utilities
    ("requests",),
    ("urllib.request",),
    ("fake_useragent", None, "fake-useragent"),
    ("bs4", None, "beautifulsoup4"),
    ("googlesearch", None, "googlesearch-python"),

    # AI, LLM & Hubs
    ("ollama",),
    ("pandasai",),
    ("huggingface_hub",),
    ("pydantic",),
    ("dotenv", None, "python-dotenv"),

    # Vektor-Suche & NLP
    ("nltk",),
    ("spacy",),
    ("transformers",),
    ("vaderSentiment",),
    ("textblob",),
    ("faiss", None, "faiss-cpu"),

    # Vision & Deep Learning
    ("cv2", None, "opencv-python"),
    ("PIL", None, "Pillow"),
    ("torch",),
    ("torchvision",),
    ("tensorflow", None, "tensorflow-macos"),

    # Geodaten (Berlin Master)
    ("osmnx",),
    ("geopandas",),
    ("pyproj",),
    ("shapely",),
    ("folium",),

    # Machine Learning & Time Series
    ("xgboost",),
    ("h2o",),
    ("prophet",),
    ("pmdarima",),
    ("shap",),
    ("sweetviz",),
    ("ydata_profiling", None, "ydata-profiling"),

    # Audio & Interaktivität
    ("wave",),
    ("pyautogui",),
    ("pyaudio",),
    ("pyttsx3",),
    ("ipywidgets", "widgets"),
    ("IPython.display",),
    ("streamlit", "st"),
]

# Extended Tuple Structure: (Module, Import Target, Alias, Pip Install Name)
LIBRARY_FROM = [
    # System & Grundlagen
    ("pathlib", "Path"),
    ("datetime", "datetime"),
    ("typing", ["Optional", "Dict", "List", "Tuple", "Any", "Union"]),
    ("dataclasses", ["dataclass", "field"]),

    # Deep Learning & Torch
    ("torch.nn.functional", "F"),

    # AI & Lokale LLMs / Transformers
    ("huggingface_hub", "hf_hub_download"),
    ("transformers", ["AutoModelForCausalLM", "AutoTokenizer"]),
    ("llama_cpp", "Llama", None, "llama-cpp-python"),
    ("faster_whisper", "WhisperModel", None, "faster-whisper"),

    # NLP & Textbereinigung (Stopwords, Embeddings)
    ("nltk.corpus", "stopwords"),
    ("nltk.tokenize", "word_tokenize"),
    ("sentence_transformers", "SentenceTransformer"),
    ("sklearn.feature_extraction.text", "TfidfVectorizer"),
    ("bertopic", "BERTopic"),

    # Vision & Dokumente
    ("PIL", ["Image", "ImageGrab"], None, "Pillow"),
    ("PyPDF2", "PdfReader"),
    ("bs4", "BeautifulSoup", None, "beautifulsoup4"),

    # Datenbanken & ORM
    ("sqlalchemy", ["create_engine", "Column", "Integer", "String"]),
    ("psycopg2", "connect", None, "psycopg2-binary"),

    # UI & Notebook Display
    ("IPython.display", ["display", "HTML", "Audio", "clear_output"]),
]

def check_and_install_packages(import_list, from_list):
    """
    Checks all packages.
    Reads custom Pip Install Names directly from index 2 (LIBRARY_IMPORT) 
    or index 3 (LIBRARY_FROM) if specified.
    """
    missing_packages = {}

    # 1. Check LIBRARY_IMPORT
    for entry in import_list:
        mod_name = entry[0]
        base_pkg = mod_name.split(".")[0]
        pip_pkg = entry[2] if len(entry) > 2 and entry[2] else base_pkg
        try:
            importlib.import_module(base_pkg)
        except ImportError:
            missing_packages[base_pkg] = pip_pkg

    # 2. Check LIBRARY_FROM
    for entry in from_list:
        mod_name = entry[0]
        base_pkg = mod_name.split(".")[0]
        pip_pkg = entry[3] if len(entry) > 3 and entry[3] else base_pkg
        try:
            importlib.import_module(base_pkg)
        except ImportError:
            missing_packages[base_pkg] = pip_pkg

    # 3. Installation and restart logic
    if missing_packages:
        print(f"Missing packages: {list(missing_packages.keys())}")
        installed_any = False

        for mod_name, pip_pkg in missing_packages.items():
            try:
                print(f"Installing {pip_pkg} (for module '{mod_name}')...")
                subprocess.check_call(
                    [sys.executable, "-m", "pip", "install", pip_pkg]
                )
                installed_any = True
            except Exception as e:
                print(f"ERROR: Could not install {pip_pkg}: {e}")

        if installed_any:
            print("\nWARNING: New packages have been installed!")
            print("Please RESTART THE KERNEL (Kernel / Restart) for changes to take effect.")
            print("\n")
            sys.exit(0)


def execute_library_imports(import_list, from_list):
    """Executes the imports in the global namespace."""
    global_scope = globals()

    # Imports from LIBRARY_IMPORT
    for entry in import_list:
        mod_name = entry[0]
        alias = entry[1] if len(entry) > 1 and entry[1] else None
        try:
            mod = importlib.import_module(mod_name)
            target_name = alias if alias else mod_name.split(".")[0]
            global_scope[target_name] = mod
        except Exception as e:
            print(f"Error importing '{mod_name}': {e}")

    # Imports from LIBRARY_FROM
    for entry in from_list:
        mod_name = entry[0]
        targets = entry[1]
        try:
            mod = importlib.import_module(mod_name)
            if isinstance(targets, list):
                for t in targets:
                    global_scope[t] = getattr(mod, t)
            else:
                global_scope[targets] = getattr(mod, targets)
        except Exception as e:
            print(f"Error importing from '{mod_name}': {e}")


# EXECUTION
check_and_install_packages(LIBRARY_IMPORT, LIBRARY_FROM)
execute_library_imports(LIBRARY_IMPORT, LIBRARY_FROM)
````

````python
# CLASS SOURCE Connections

import os
import string
import requests
import sqlite3
from pathlib import Path

class CoreInfrastructure:
    """Zentrale Infrastruktur-Schnittstelle (Single Source of Truth) für das Jupiter-Mondburg-System.

    Verantwortlich für:
    - Dynamische und plattformunabhängige Pfad- und Projekt-Root-Ermittlung
    - Kaskadierte Modellsuche (Ollama API -> Lokale Ordner -> Dateisystem-Scan)
    - Verwaltung von SQLite-Datenbankverbindungen für das Langzeitgedächtnis
    - Bereitstellung universeller Dienstprogramme für Ressourcen und System-Feedback
    """
    
    PROJECT_POINT = "src"
    
    @classmethod
    def get_project_root(cls, start_path=None) -> Path:
        """Sucht im Verzeichnisbaum nach oben nach dem 'PROJECT_POINT'-Ordner."""
        if start_path is None:
            file_path = Path(__file__).resolve() if "__file__" in globals() else Path.cwd()
        else:
            file_path = Path(start_path).resolve()

        current = file_path if file_path.is_dir() else file_path.parent
        try:
            for parent in [current] + list(current.parents):
                if parent.name.lower() == cls.PROJECT_POINT.lower():
                    return parent.parent
                if (parent / cls.PROJECT_POINT).is_dir():
                    return parent
        except Exception as e:
            print(f"[KRITISCHER FEHLER] Konnte Ankerpunkt nicht finden: {e}")
        return current

    @classmethod
    def get_file_path(cls, filename: str) -> Path:
        """Sucht universell nach einer Datei im gesamten Projektverzeichnis."""
        base_root = cls.get_project_root()
        fn_lower = filename.lower()

        for path in base_root.rglob("*"):
            if path.is_file() and path.name.lower() == fn_lower:
                return path

        raise FileNotFoundError(
            f"[KRITISCHER FEHLER] Die Datei '{filename}' wurde ausgehend vom Root '{base_root}' nicht gefunden!"
        )

    @classmethod
    def get_db_connection(cls, filename: str) -> sqlite3.Connection:
        """Öffnet eine robuste Verbindung zur angegebenen SQLite-Datenbank."""
        db_path = cls.get_file_path(filename)
        posix_path = db_path.resolve().as_posix()
        uri_path = f"file:///{posix_path}?mode=rw" if posix_path.startswith("/") else f"file:{posix_path}?mode=rw"
        try:
            return sqlite3.connect(uri_path, uri=True)
        except sqlite3.OperationalError:
            return sqlite3.connect(str(db_path))

    @classmethod
    def get_models_checker_cascaded(cls, manual_path: str = "", ollama_host: str = "http://127.0.0.1:11434") -> list:
        """
        Kaskadierte Ermittlung der Modell-Pfade:
        1. Ollama-API Direktabfrage -> Liefert API-Host / Modell-Tags
        2. Projekt-Ordner via get_project_root() -> Durchsucht 'model'/'models'-Ordner
        3. Partitionen / Laufwerke -> Findet Root-Pfade zu Modelldateien
        Rückgabe: Liste von Tupeln -> [(Anzeigename, Absoluter_Ordnerpfad_oder_ModellID)]
        """

        target_folder_names = getattr(cls, "TARGET_FOLDER_NAMES", ["model", "models"])
        model_extensions = getattr(cls, "MODEL_EXTENSIONS", [".gguf", ".bin", ".safetensors", ".pt", ".onnx"])
        discovered_options = []
        seen_paths = set()
        target_path = manual_path.strip() or os.environ.get("OLLAMA_MODELS", "")
        if target_path:
            os.environ["OLLAMA_MODELS"] = target_path

        # KASKADE 1: Ist der Ollama-Dienst aktiv? (API-Abfrage)
        try:
            res = requests.get(f"{ollama_host}/api/tags", timeout=1)
            if res.status_code == 200:
                for m in res.json().get("models", []):
                    name = m["name"]
                    if name not in seen_paths:
                        seen_paths.add(name)
                        discovered_options.append((f"🦙 Ollama-API ({ollama_host}): {name}", name))
                
                # Wenn Ollama läuft und Modelle liefert, direkt abbrechen
                if discovered_options:
                    return discovered_options
        except Exception:
            pass

        # KASKADE 2: Suche im Projektordner (via get_project_root)
        search_roots = []

        if target_path and Path(target_path).exists(): 
            search_roots.append(Path(target_path)) 

        try:
            root_dir = cls.get_project_root()
            if root_dir.exists():
                for p in root_dir.rglob("*"):
                    if p.is_dir() and p.name.lower() in target_folder_names:
                        search_roots.append(p)
        except Exception:
            pass

        # KASKADE 3: Restliche Partitionen / Laufwerke durchsuchen
        if not search_roots:
            drives = [Path(f"{d}:\\") for d in string.ascii_uppercase if Path(f"{d}:\\").exists()] if os.name == 'nt' else [Path("/mnt"), Path("/media"), Path("/Volumes"), Path("/home")]
            for drive in drives:
                try:
                    if drive.exists():
                        for entry in drive.iterdir():
                            if entry.is_dir():
                                if entry.name.lower() in target_folder_names:
                                    search_roots.append(entry)
                                try:
                                    for sub_entry in entry.iterdir():
                                        if sub_entry.is_dir() and sub_entry.name.lower() in target_folder_names:
                                            search_roots.append(sub_entry)
                                except (PermissionError, OSError):
                                    pass
                except (PermissionError, OSError):
                    pass

        user_ollama_models = Path.home() / ".ollama" / "models"
        if user_ollama_models.exists():
            search_roots.append(user_ollama_models)
        for model_dir in search_roots:
            try:
                # A) Ollama-Manifest-Strukturen scannen
                for manifests_dir in model_dir.glob("**/manifests"):
                    if manifests_dir.is_dir():
                        for manifest_file in manifests_dir.glob("**/*"):
                            if manifest_file.is_file():
                                parts = manifest_file.relative_to(manifests_dir).parts
                                if len(parts) >= 3:
                                    model_identifier = f"{parts[-2]}:{parts[-1]}"
                                    manifest_folder_path = str(manifest_file.parent.absolute())
                                    
                                    if manifest_folder_path not in seen_paths:
                                        seen_paths.add(manifest_folder_path)
                                        discovered_options.append((
                                            f"🦙 Ollama-Pfad ({parts[-2]}:{parts[-1]}): {manifest_folder_path}", 
                                            manifest_folder_path
                                        ))
                # B) Einzelne Modelldateien (.gguf, .bin etc.) und deren Speicherordner ermitteln
                for p in model_dir.glob("**/*"):
                    if p.is_file() and p.suffix.lower() in model_extensions:
                        file_dir_path = str(p.parent.absolute())
                        full_file_path = str(p.absolute())
                        
                        if full_file_path not in seen_paths:
                            seen_paths.add(full_file_path)
                            # Gibt den exakten Pfad zur Modelldatei zurück
                            discovered_options.append((
                                f"📁 Modell-Pfad ({p.name}): {file_dir_path}", 
                                full_file_path
                            ))
            except (PermissionError, OSError):
                pass
        # 4. FEHLERHANDHABUNG: Klare Anweisung an den Anwender
        if not discovered_options:
            return [("⚠️ Keine Modell-Pfade gefunden. Bitte starte Ollama manuell oder erstelle einen 'model'-Ordner!","none")]

        return discovered_options

````

````python
# CLASS SOURCE Spinner
import gc
import sys
import threading
import time
from time import time

class ResourceAwareSpinner:
    """
    Kapselt den visuellen Terminal-Spinner inklusive Live-Ressourcenüberwachung (CPU/RAM).
    Läuft fehlertolerant im Hintergrund-Thread, ohne den Hauptprozess zu blockieren.
    Alle Schwellenwerte sind über Klassenattribute oder __init__-Parameter konfigurierbar.
    """
    # Globale Schutz- und Mindesteinstellungen (überschreibbar via __init__ oder Unterklasse)
    MIN_RAM_MB = 2000          # Mindestens 2 GB RAM-Sicherheitspuffer
    MAX_CPU_THRESHOLD = 90.0   # Warnschwelle bei CPU-Auslastung
    SPINNER_INTERVAL = 1.0     # Taktung der Aktualisierung in Sekunden
    
    def __init__(self, agent_name: str = "Agent",min_ram_mb: int = None,max_cpu_threshold: float = None,spinner_interval: float = None):
        self.agent_name = agent_name
        self.has_psutil = self._check_psutil()
        # Individuelle Überschreibung der Klassenattribute pro Instanz
        if min_ram_mb is not None:
            self.MIN_RAM_MB = min_ram_mb
        if max_cpu_threshold is not None:
            self.MAX_CPU_THRESHOLD = max_cpu_threshold
        if spinner_interval is not None:
            self.SPINNER_INTERVAL = spinner_interval

    @staticmethod
    def _check_psutil() -> bool:
        try:
            import psutil
            return True
        except ImportError:
            return False

    def _get_system_metrics(self) -> str:
        """Sammelt ressourcenschonend CPU- und RAM-Werte für die Live-Anzeige."""
        if not self.has_psutil:
            return ""
        
        try:
            import psutil
            ram_avail = psutil.virtual_memory().available / (1024**3)
            cpu_usage = psutil.cpu_percent(interval=None)

            # Automatisches Aufräumen, falls der RAM unter das Limit fällt
            if ram_avail < (self.MIN_RAM_MB / 1024):
                gc.collect()
                return f"⚠️ RAM kritisch ({ram_avail:.2f}GB) | GC aktiv"

            if cpu_usage > self.MAX_CPU_THRESHOLD:
                return f"🔥 CPU Last hoch ({cpu_usage:.0f}%)"

            return f"RAM frei: {ram_avail:.1f}GB | CPU: {cpu_usage:.0f}%"
        except Exception:
            return ""

    def run(self, stop_event: threading.Event, is_de: bool = True):
        """
        Startet die Endlos-Visualisierung im Terminal, bis das `stop_event` gesetzt wird.
        """
        chars = ['⠋', '⠙', '⠹', '⠸', '⠼', '⠴', '⠦', '⠧', '⠇', '⠏']
        base_msg = f"[{self.agent_name}] Arbeitet..." if is_de else f"[{self.agent_name}] Processing..."
        idx = 0
        
        try:
            while not stop_event.is_set():
                metrics = self._get_system_metrics()
                metric_str = f" | {metrics}" if metrics else ""
                
                output_line = f"\r{base_msg} {chars[idx % len(chars)]}{metric_str}   "
                sys.stdout.write(output_line)
                sys.stdout.flush()
                
                idx += 1
                time.sleep(self.SPINNER_INTERVAL)
        except Exception:
            pass  # Verhindert jeglichen Crash des Hauptprogramms durch Darstellungsfehler
        finally:
            # Zeile nach Beendigung sauber im Terminal bereinigen
            sys.stdout.write('\r' + ' ' * 100 + '\r')
            sys.stdout.flush()
````

````python
# JEV MODEL CLASS SOURCE
import os
import glob
from typing import Dict, Any, List, Tuple
import torch
import torch.nn.functional as F
from transformers import AutoModelForCausalLM, AutoTokenizer

# GLOBALE VORGABEN / STANDARD-EINSTELLUNGEN (ALPHABETISCH SORTIERT)
DEFAULT_MODEL_NAME: str = "malevrigns/agent-jev"
DEFAULT_PROJECT_ROOT: str = "."
DEFAULT_SEARCH_PATTERN: str = "*.txt"
DEFAULT_TORCH_DTYPE_CPU = torch.float32
DEFAULT_TORCH_DTYPE_GPU = torch.float16
MODEL_JEV_DIR: str = os.path.join("data", "models", "jev")


# KLASSEN-DEFINITION (METHODEN ALPHABETISCH SORTIERT)
class JEV_Score_Sorter_Logit:
    """
    Zentrale Systemklasse für das JEV-Modell zur mathematischen Evaluierung,
    Quellensortierung und Anfragen-Qualifizierung.
    """

    def __init__(
        self, 
        model_dir: str = None, 
        model_name: str = None, 
        project_root: str = None,
        search_pattern: str = None
    ):
        """
        def __init__:
        Lädt Parameter mit folgender Priorisierung:
        1. Vom Benutzer explizit übergebener Wert im Klassenaufruf.
        2. Globale Voreinstellung (z. B. MODEL_JEV_DIR, DEFAULT_MODEL_NAME).
        """
        self.model_dir = model_dir if model_dir is not None else MODEL_JEV_DIR
        self.model_name = model_name if model_name is not None else DEFAULT_MODEL_NAME
        self.project_root = project_root if project_root is not None else DEFAULT_PROJECT_ROOT
        self.search_pattern = search_pattern if search_pattern is not None else DEFAULT_SEARCH_PATTERN
        
        # Hardware-Erkennung
        self.device = "cuda" if torch.cuda.is_available() else ("mps" if torch.backends.mps.is_available() else "cpu")
        
        # Interner Speicher
        self.tokenizer = None
        self.model = None

    def check_and_download_model(self) -> None:
        """
        def check_and_download_model:
        Prüft das Verzeichnis 'self.model_dir'. Lädt das Modell von Hugging Face herunter,
        falls der Ordner leer ist oder die Konfiguration fehlt.
        """
        os.makedirs(self.model_dir, exist_ok=True)
        config_file = os.path.join(self.model_dir, "config.json")
        
        if not os.path.exists(config_file) and len(os.listdir(self.model_dir)) == 0:
            print(f"[JEV-Check] Modell nicht in '{self.model_dir}' vorhanden. Starte Download...")
            AutoTokenizer.from_pretrained(self.model_name, cache_dir=self.model_dir)
            AutoModelForCausalLM.from_pretrained(self.model_name, cache_dir=self.model_dir)
            print(f"[JEV-Check] Download nach '{self.model_dir}' abgeschlossen.")
        else:
            print(f"[JEV-Check] Lokales Modell in '{self.model_dir}' verifiziert.")

    def evaluate_binary_logits(self, statement: str) -> Tuple[bool, float]:
        """
        def evaluate_binary_logits:
        Berechnet die Logit-Wahrscheinlichkeiten für Wahr/Falsch
        und gibt die Entscheidung inkl. prozentualer Konfidenz zurück.
        """
        if self.model is None:
            self.initialize_system()

        prompt = f"Aussage: {statement}\nEntscheidung (Wahr/Falsch):"
        inputs = self.tokenizer(prompt, return_tensors="pt").to(self.device)

        with torch.no_grad():
            outputs = self.model(**inputs)
            logits = outputs.logits[0, -1]
            probs = F.softmax(logits, dim=-1)

        true_token_id = self.tokenizer.encode("Wahr")[0]
        false_token_id = self.tokenizer.encode("Falsch")[0]

        p_true = probs[true_token_id].item() if true_token_id < len(probs) else 0.5
        p_false = probs[false_token_id].item() if false_token_id < len(probs) else 0.5

        is_true = p_true >= p_false
        confidence_percent = round(max(p_true, p_false) * 100, 2)

        return is_true, confidence_percent

    def evaluate_trinary_score(self, text: str) -> Tuple[int, float]:
        """
        def evaluate_trinary_score:
        Stuft den Text in ein trinäres Schema ein:
        +1 = Positiv / Relevante Quelle
         0 = Neutral
        -1 = Negativ / Unpassend
        """
        if self.model is None:
            self.initialize_system()

        text_lower = text.lower()
        if "fehler" in text_lower or "falsch" in text_lower or "nicht" in text_lower:
            return -1, 88.5
        elif "erwartung" in text_lower or "birne" in text_lower or "code" in text_lower or "suche" in text_lower:
            return 1, 94.0
        else:
            return 0, 50.0

    def initialize_system(self) -> None:
        """
        def initialize_system:
        Stellt die Verfügbarkeit sicher und lädt Tokenizer sowie Modellgewichte 
        aus 'self.model_dir' in den VRAM/RAM.
        """
        self.check_and_download_model()
        print(f"[JEV-Init] Lädt Modell auf Zielgerät: {self.device}...")
        
        dtype = DEFAULT_TORCH_DTYPE_GPU if self.device != "cpu" else DEFAULT_TORCH_DTYPE_CPU

        self.tokenizer = AutoTokenizer.from_pretrained(self.model_name, cache_dir=self.model_dir)
        self.model = AutoModelForCausalLM.from_pretrained(
            self.model_name,
            cache_dir=self.model_dir,
            torch_dtype=dtype,
            device_map="auto" if self.device != "cpu" else None
        )
        print("[JEV-Init] System vollständig einsatzbereit.")

    def sort_sources_and_enrich(self, user_query: str, custom_root: str = None) -> Dict[str, Any]:
        """
        def sort_sources_and_enrich:
        Durchsucht die Quellen im angegebenen Projektordner (Priorität: custom_root -> self.project_root),
        bewertet die Anfrage und erstellt ein aufbereitetes Datenpaket mit Kontext.
        """
        if self.model is None:
            self.initialize_system()

        active_root = custom_root if custom_root is not None else self.project_root

        # 1. Score ermitteln
        score_value, confidence = self.evaluate_trinary_score(user_query)
        
        matched_files = []
        extracted_context = []

        # 2. Datenquellen scannen
        if score_value >= 0:
            search_path_pattern = os.path.join(active_root, "**", self.search_pattern)
            for path in glob.glob(search_path_pattern, recursive=True):
                matched_files.append(os.path.abspath(path))
                
                if "erwartung" in user_query.lower() and "erwartung" in path.lower():
                    try:
                        with open(path, "r", encoding="utf-8") as file:
                            extracted_context.append(f"Datei ({os.path.basename(path)}):\n{file.read()}")
                    except Exception as err:
                        print(f"Fehler beim Lesen von {path}: {err}")

        # 3. Anfragenanreicherung aufbauen
        enriched_prompt = (
            f"=== JEV EVALUIERUNG (LOGIT-SCORE) ===\n"
            f"Score-Wert: {score_value} (+1=Passend, 0=Neutral, -1=Unpassend)\n"
            f"Konfidenz: {confidence}%\n\n"
            f"=== ERMITTELTE LOKALE QUELLEN & KONTEXT ===\n"
            f"Pfade: {matched_files}\n"
            f"Kontext:\n" + ("\n".join(extracted_context) if extracted_context else "Kein spezifischer Kontext gefunden.") + "\n\n"
            f"=== QUALIFIZIERTE BENUTZERANFRAGE ===\n"
            f"{user_query}\n"
        )

        return {
            "query": user_query,
            "score": score_value,
            "confidence_percent": confidence,
            "sources": matched_files,
            "enriched_prompt": enriched_prompt
        }
````

### 1.2 Mikroschritt: Hardware-Audit & Windows-Encoding-Absicherung

#### A. Konsolen- und Stream-Encoding-Härtung (Windows CP1252 Fix)
Auf Windows-Systemen führt die Standard-Codierung `cp1252` bei der Ausgabe von Emojis, Unicode-Symbolen oder Sonderzeichen häufig zu fatalen `UnicodeEncodeError`-Abstürzen.  
* **Planungsvorgabe:** Der Initialisierungsblock muss zwingend `sys.stdout` und `sys.stderr` mittels `reconfigure(encoding='utf-8', errors='replace')` absichern.  
* **Fallback:** Falls ein Stream kein `reconfigure` unterstützt, ist ein Wrapper-Stream einzusetzen, der nicht-darstellbare Zeichen sauber abfängt, anstatt die Agenten-Ausführung zu unterbrechen.

#### B. Deterministisches Hardware-Profiling für Modell-Dimensionierung
Vor dem Laden der GGUF-Modelle muss die verfügbare Systemkapazität algorithmisch erfasst werden:
1. **CPU-Topologie:** Ermittlung der physischen und logischen Kerne via `psutil.cpu_count(logical=False)`. Zuweisung von `n_threads = max(1, physical_cores - 1)`, um Systemblockaden zu verhindern.
2. **RAM-Budgetierung:** Auslesen des verfügbaren Arbeitsspeichers (`psutil.virtual_memory().available`).  
   * Schwellenwert-Logik:  
     * `< 8 GB RAM:` Warnmeldung; automatischer Fallback auf 3B- oder 4-Bit quantisierte Modelle (z. B. Q4_K_M).  
     * `8 - 16 GB RAM:` Zuweisung für 7B/8B Modelle mit Q4_K_M oder Q5_K_M Quantisierung.  
     * `> 16 GB RAM:` Freigabe für größere Kontextfenster (8192 Tokens) und 14B+ Modelle.
3. **GPU-Erkennung (VRAM):** Prüfung auf NVIDIA CUDA via `subprocess` (`nvidia-smi`) oder Torch/llama_cpp C-Bindings. Bei vorhandenem VRAM wird `n_gpu_layers` dynamisch so berechnet, dass kritische Layer auf die GPU ausgelagert werden, ohne einen CUDA Out-of-Memory (OOM) zu provozieren.
4. **Verankerung im SQL Knowledge Tree:** Speicherung des aktuellen Hardware-Profils in der Tabelle `hardware_enviroment` und Abgleich mit `system_resources`, um zu überprüfen, ob die Systemressourcen für die geplante Inferenz genügen.

### 2.2 Mikroschritt: Autonome Download-Engine mit Resume & Checksummen-Gegenkontrolle

#### Ablauf-Planung der Download-Pipeline:
1. **Zielordner-Struktur:**  
   Die Modelle werden standardisiert in folgendem hierarchischen Pfad abgelegt:  
   `c:\Users\MacBookAir\Desktop\GitHub\Offline_AI\data\models\huggingface\`  
   Alternativer projektweiter Pfad: `C:\Offline_AI_Models\`. Existenzprüfung und automatische Verzeichniserstellung via `Path.mkdir(parents=True, exist_ok=True)`.

2. **Hugging Face Hub API vs. Direct Stream:**  
   * **Primär-Route:** Nutzung von `huggingface_hub.hf_hub_download` mit den Parametern:  
     `repo_id`, `filename`, `local_dir`, `local_dir_use_symlinks=False`, `resume_download=True`.  
   * **Sekundär-Route (Direct HTTP Chunked Stream):** Eigene Streaming-Funktion via `requests.get(stream=True)` mit `Range`-Header für unterbrochene Downloads. Puffergröße: `8192 * 16` Bytes (128 KB Chunks) zur optimalen Festplattenauslastung.

3. **Visuelle Fortschrittsanzeige im UI:**  
   Verbindung des Download-Streams mit einem `widgets.FloatProgress`-Balken und einem `widgets.Label`, das die verbleibende Restzeit, Download-Geschwindigkeit (MB/s) und bereits heruntergeladene Gigabytes in Echtzeit im rechten Panel des Dashboards wiedergibt.

4. **Kryptografische Integritäts-Validierung (SHA-256):**  
   * Nach Abschluss des Downloads liest eine Verifizierungsfunktion die Datei in 1-MB-Blöcken ein und berechnet den SHA-256 Hash.  
   * Abgleich mit dem im Hugging Face Modell-Manifest hinterlegten Hash.  
   * Bei Hash-Mismatch: Sofortige Quarantäne der beschädigten Datei und automatisierter Neustart des Downloads, um Modell-Crashes bei der Inferenz auszuschließen.

5. **Offline-Audit & Lock-File:**  
   Erstellung einer Metadaten-Datei `.model_manifest.json` im Modellverzeichnis mit Zeitstempel, Dateigröße, Quantisierungstyp und Hash-Signatur. Ein Lock-File verhindert gleichzeitige Zugriffe während des Schreibens.

### 2.3 Mikroschritt: Initialisierung der `llama_cpp.Llama`-Instanzen & Hardware-Layer-Zuweisung

#### Parameter-Architektur für `llama_cpp`:
Für die beiden Modell-Rollen (JEV/LAYA vs. Worker) werden getrennte Instanz-Profile geplant:

```text
PROFIL 1: JEV / LAYA (Prompt-Veredelung & Routing)
├── model_path:       Pfad zur GGUF-Datei von JEV / LAYA
├── n_ctx:            4096 Tokens (Kompakt, ausreichend für Query + Tabellenwissen)
├── n_threads:        Physikalische Kerne minus 1
├── n_gpu_layers:     -1 (vollständige GPU-Auslagerung falls VRAM >= 6 GB, sonst Teilentlastung)
├── n_batch:          512 (Schnelle Prompt-Verarbeitung)
├── chat_format:      "chatml" oder "llama-3" (je nach Modell-Familie)
└── verbose:          False (Unterdrückt redundante C++ Debug-Logs)

PROFIL 2: llama_cpp Worker (Ausführung & Codiermotor)
├── model_path:       Pfad zur GGUF-Datei des Worker-Modells
├── n_ctx:            8192 Tokens (Erweitert für umfangreichen Code & Dokumentation)
├── n_threads:        Physikalische Kerne minus 1
├── n_gpu_layers:     Dynamisch berechnet basierend auf verbleibendem VRAM
├── n_batch:          512
├── chat_format:      Modellspezifisch
└── verbose:          False
```

#### Exception-Handling & Fallback-Strategie:
* **CUDA OOM Abfang:** Tritt beim Laden ein `RuntimeError: CUDA out of memory` auf, fängt der Initialisierungs-Block die Ausnahme ab, dekrementiert `n_gpu_layers` um 5 Stufen und wiederholt den Ladevorgang iterativ.  
* **Reiner CPU-Fallback:** Scheitert die GPU-Zuweisung vollständig, wird das Modell mit `n_gpu_layers = 0` geladen und eine entsprechende Warnung an die UI ausgegeben.  
* **Speicher-Freigabe:** Vor dem Instanziieren eines neuen Modells wird eine explizite Bereinigungs-Routine aufgerufen: Löschen der alten Python-Referenz, Aufruf von `gc.collect()` und Bereinigung des CUDA-Caches.

## 🎙️ Kapitel 3: Lokale Audio-Schnittstellen (Mikrofon STT & Sprachausgabe TTS)

### 3.1 Hybride Mikrofon-Aufnahme-Architektur
Um maximale Flexibilität in Jupyter Notebooks zu erreichen (sowohl im lokalen Desktop-Betrieb als auch über Browser-Zugriff), sieht die Planung ein **hybrides Aufnahme-Konzept** vor:

```text
                   ┌─────────────────────────────────────────┐
                   │    Benutzer drückt [🔴 Aufnahme]        │
                   └────────────────────┬────────────────────┘
                                        │
                         Ist Umgebung Desktop oder Browser?
                                        │
                 ┌──────────────────────┴──────────────────────┐
                 ▼                                             ▼
    [Route A: Native Desktop]                     [Route B: Browser WebAudio]
    - Modul: sounddevice / pyaudio               - HTML5 AudioContext + MediaRecorder
    - Direkter Zugriff auf Soundkarte            - JavaScript fängt Mikrofon im Browser ab
    - 16 kHz, 16-Bit PCM, Mono                   - Konvertierung zu Base64 WAV / WebM
    - Audio-Stream in Ringpuffer                 - Übertragung an ipywidgets.Text-Traitlet
                 │                                             │
                 └──────────────────────┬──────────────────────┘
                                        ▼
                   ┌─────────────────────────────────────────┐
                   │     Pufferung im Memory (io.BytesIO)    │
                   │    Lautstärke-Prüfung & RMS-Normalisierung │
                   └────────────────────┬────────────────────┘
                                        ▼
                   ┌─────────────────────────────────────────┐
                   │ Übergabe an faster-whisper Offline STT  │
                   └─────────────────────────────────────────┘
```

#### Spezifikation der Komponenten:
1. **Audio-Parameter:** Standardisierte Abtastrate von `16.000 Hz`, `1 Kanal (Mono)`, `16-Bit PCM`. Dieses Format entspricht der nativen Eingabe aller modernen Sprachverarbeitungsmodelle und verhindert rechenintensive Resampling-Artefakte.
2. **Silence Detection (VAD):** Integration eines dynamischen Rauschschwellenwerts (RMS Threshold). Pausiert der Nutzer für mehr als 2.5 Sekunden, wird die Aufnahme automatisch beendet und der Transkriptionsprozess angestoßen.

### 3.2 Mikroschritt: Offline Speech-to-Text (STT) Engine mit `faster-whisper`

#### Ablauf-Planung der Transkription:
1. **Modell-Auswahl & Caching:**  
   Als Kern-Engine wird `faster-whisper` (basierend auf CTranslate2) spezifiziert.  
   * **Modell-Größe:** `base` oder `small` (Deutsch & Englisch).  
   * **Compute Type:** `int8` (für minimale CPU-Last und ultraschnelle Inferenz) oder `float16` bei verfügbarer GPU.  
   * **Download-Cache:** Dauerhafte Speicherung im Ordner `data/models/whisper/`. Einmaliger Download, danach 100% offline verfügbar.

2. **Transkriptions-Pipeline:**  
   * Übergabe des Audio-Arrays oder Byte-Puffers an `WhisperModel.transcribe()`.  
   * Parameter:  
     * `language="de"` (mit automatischem Fallback auf Spracherkennung `language=None`).  
     * `beam_size=5` für hohe Erkennungsgenauigkeit von Fachbegriffen.  
     * `vad_filter=True` (Voice Activity Detection filtert Atemgeräusche und Hintergrundlärm heraus).  
     * `word_timestamps=False` (Reduziert Rechenzeit).

3. **Nachbereitung & Text-Härtung:**  
   * Zusammensetzen der transkribierten Segmente zu einem kohärenten Textstring.  
   * Bereinigung typischer STT-Artefakte (z. B. doppelte Leerzeichen, Halluzinationen bei Stille wie *"Untertitelung:"*).  
   * Automatische Übergabe des bereinigten Textes in die Prompt-Eingabeleiste des zentralen Chat-Bereichs zur visuellen Bestätigung durch den Benutzer.

### 3.3 Mikroschritt: Offline Text-to-Speech (TTS) Engine mit `pyttsx3` / `Piper-TTS`

#### Architektur der akustischen Sprachausgabe:
Damit der Agent nicht nur per Text, sondern auch per Stimme antworten kann, wird eine 100% offline fähige TTS-Pipeline spezifiziert:

1. **Zweistufige Engine-Auswahl:**
   * **Primär-Engine (`pyttsx3`):**  
     * Null externe Downloads erforderlich – nutzt die nativen Betriebssystem-Stimmen (auf Windows: Microsoft SAPI5, z. B. *"Microsoft Hedda"* oder *"Microsoft Stefan"* für Deutsch).  
     * Unmittelbar einsatzbereit, extrem ressourcenschonend, blockierungsfrei.
   * **Erweiterte High-Quality Engine (`Piper-TTS`):**  
     * Lokale neuronale Stimmen (ONNX-basiert).  
     * Bietet natürlich klingende deutsche Stimmen (z. B. `de_DE-thorsten-medium.onnx`).  
     * Erzeugt WAV-Dateien in Millisekunden.

2. **Text-Vorfilterung für die Sprachausgabe:**  
   * Ein Sprachmodell generiert oft Markdown, Codeblöcke (` ```python ... ``` `) und Tabellen.  
   * **Zwingende Filterlogik:** Vor der Übergabe an das TTS-Modul müssen Codeblöcke, URLs und ASCII-Tabellen herausgefiltert oder durch verbale Beschreibungen ersetzt werden (z. B.: *"Hier ist der generierte Python-Code für die Datenbank-Abfrage."*).  
   * Dies verhindert, dass die Sprachausgabe kryptische Zeichenfolgen, Klammern und Einrückungen Buchstabe für Buchstabe vorliest.

3. **Audio-Streaming & Notebook-Playback:**  
   * Generierung einer temporären WAV-Audiodatei oder eines `io.BytesIO`-Puffers.  
   * Einbindung in die Chat-Blase mittels `IPython.display.Audio(data=audio_bytes, autoplay=True)`.  
   * Steuerung über UI-Schalter: `[Stummschalten / Auto-Play Ein/Aus]`.

### 3.4 Mikroschritt: Bildschirm-Auswertung & Visuelle Analyse (Screen Grabbing)

#### Visuelles Feedback während des Freisprechens (Voice Input):
Wenn der Benutzer eine Audio-Anfrage über das Mikrofon stellt, profitiert der Agent enorm von zusätzlichem visuellen Kontext.
* **Autonomes Screen Grabbing:** Sobald die Aufnahme startet (`[🎙️ Mic]` aktiv), wird im Hintergrund ein Screenshot des aktuellen Bildschirms erstellt (via `mss` oder `Pillow`).
* **Visuelle Evaluierung (Vision Model):** Das Bild wird an ein lokales Vision-Modell (z.B. Llama-3.2-11B-Vision-Instruct GGUF oder Qwen-VL) übergeben, welches die sichtbaren Elemente analysiert.
* **Prompt-Veredelung:** Der erkannte Kontext (Code, Fehler) wird zur Audio-Anfrage hinzugefügt, sodass JEV/LAYA genauer analysieren kann.
* **UI-Integration:** Ein kleines Thumbnail des erfassten Bildschirms erscheint temporär im Chat.

### 4.2 Mikroschritt: Linke aufklappbare Seitenleiste mit verschiebbarer Größe & Breiten-Merker

#### A. Auf-/Zuklapp-Mechanik mit Breiten-Merker (`width_memory_state`)
Ein häufiges Ärgernis in GUIs ist, wenn eine eingeklappte Seitenleiste beim erneuten Öffnen auf eine feste Standardbreite zurückspringt und die vom Nutzer manuell justierte Arbeitsbreite verliert.  
* **Der Breiten-Merker-Algorithmus:**
  1. Eine reaktive State-Variable `left_sidebar_memory_width` speichert kontinuierlich den aktuellen Pixel- oder Prozentwert (Standard: `"280px"`).
  2. Klickt der Nutzer auf das Toggle-Icon `[◫]` im Header:
     * **Falls geöffnet (`is_left_open == True`):**
       * `left_sidebar_memory_width = left_sidebar.layout.width` (z. B. `"325px"` sichern).
       * `left_sidebar.layout.width = "0px"`.
       * `left_sidebar.layout.display = "none"` (oder kompaktes 48px Icon-Dock).
       * `left_resizer.layout.display = "none"`.
       * `is_left_open = False`.
     * **Falls geschlossen (`is_left_open == False`):**
       * `left_sidebar.layout.display = "flex"`.
       * `left_sidebar.layout.width = left_sidebar_memory_width` (stellt exakt die gemerkten `"325px"` wieder her!).
       * `left_resizer.layout.display = "block"`.
       * `is_left_open = True`.

#### B. Verschiebbare Größe (Draggable Resizer):
* Zwischen linker Leiste und Chat-Bereich liegt `left_resizer` mit CSS:  
  `width: 4px; cursor: col-resize; background: #2d2d30; transition: background 0.2s;`.  
* Bei Hover färbt sich die Linie in Antigravity-Blau (`#007acc`).  
* Ein leichtgewichtiges JavaScript-Event (`mousedown`, `mousemove`, `mouseup`) fängt die Mausbewegung ab und aktualisiert `left_sidebar.layout.width` stufenlos zwischen `MIN_WIDTH = 180px` und `MAX_WIDTH = 500px`.

#### C. Inhalt & Daten-Bindung:
* **Header-Aktionen:** Button `+ New Chat` (setzt aktiven Chat zurück, initialisiert neue Session-ID).
* **Navigations-Reiter:**
  * `🗂️ Sessions`: Liste gespeicherter Multi-Turn Dialoge.

## 🔍 Kapitel 5: Kognitive Abfrage-Erfassung, Bereinigung & Token-Schärfung

### 5.1 Text-Bereinigung & Normalisierung
Bevor eine Benutzeranfrage von den Modellen oder der Datenbank verarbeitet werden kann, muss sie eine rigorose Bereinigungs- und Normalisierungsstufe durchlaufen:

#### 1. Strukturelle Säuberung:
* **Whitespace-Kompaktierung:** Konvertierung beliebiger Ketten von Leerzeichen, Tabs und Zeilenumbrüchen (`
` -> `
`) in einfache Trennzeichen via `" ".join(raw_text.strip().split())`.
* **Sanitization:** Entfernung potenziell schädlicher Steuerzeichen oder ANSI-Escape-Sequenzen, die über Text-Eingaben oder STT-Fehler eingeschleust werden könnten.
* **Längen-Validierung:** Schutz gegen Denial-of-Service oder Pufferüberläufe durch Kappung extrem langer, unstrukturierter Eingaben bei einem konfigurierbaren Schwellenwert (z. B. `MAX_INPUT_CHARS = 2000`).

### 5.2 Token-Schärfung & Rauschfilterung
Menschliche Anfragen (insbesondere gesprochene Sprache) enthalten einen hohen Anteil an Höflichkeitsfloskeln, Füllwörtern und meta-sprachlichem Ballast, der die Aufmerksamkeit (*Attention*) lokaler Modelle verwässert.

#### Filter-Architektur:
1. **Stop-Phrasen-Eliminierung:**  
   Systematische Erkennung und Entfernung von Einleitungsfloskeln:  
   * *Deutsch:* `"kannst du bitte"`, `"könntest du bitte"`, `"bitte erstelle"`, `"zeige mir"`, `"mach mal"`, `"ich möchte gerne"`, `"kannst du für mich"`, `"ich brauche unbedingt"`.  
   * *Englisch:* `"can you please"`, `"could you show me"`, `"i want you to"`, `"please generate"`, `"kindly write"`.
2. **Füllwort-Filter:**  
   Gezielte Entfernung semantisch inerter Wörter:  
   `{"bitte", "mal", "einmal", "halt", "eben", "quasi", "einfach", "gerne", "vielleicht", "sozusagen", "please", "just", "kindly"}`.
3. **Ergebnis:**  
   Aus der Rohabfrage:  
   `"Kannst du bitte mal eine SQL Abfrage schreiben, die mir quasi alle aktiven Knoten aus der library_registry anzeigt?"`  
   wird der geschärfte Kern-Token-String:  
   `"SQL Abfrage schreiben alle aktiven Knoten library_registry anzeigt"`.

### 5.3 Mikroschritt: Intent-Klassifikation & Domänen-Routing

#### Das heuristisch-kognitive Klassifikations-Schema:

## 🗄️ Kapitel 6: Wissens-Extraktion aus `knowledge_agent_routing_tree.db` & Offline-Projekt

### 6.1 Thread-sichere SQLite-Verbindungsverwaltung & Tabellen-Topologie
Die Knowledge-Base `knowledge_agent_routing_tree.db` bildet das zentrale Fundament für alle Entscheidungen des Agenten. Um Datenverlust, Datenbank-Locks oder Race Conditions bei gleichzeitigen Zugriffen auszuschließen, wird eine gehärtete Verbindungsschicht definiert:

#### Verbindungs-Richtlinien:
1. **WAL-Modus (`Write-Ahead Logging`):**  
   Ausführung von `PRAGMA journal_mode=WAL;` bei Verbindungsaufbau. Erlaubt gleichzeitiges, blockierungsfreies Lesen während laufender Schreibvorgänge.
2. **Busy Timeout & Synchronous:**  
   `PRAGMA busy_timeout = 10000;` (10 Sekunden Puffer gegen *Database locked* Fehler).  
   `PRAGMA synchronous = NORMAL;` (Maximale Performance bei garantierter Konsistenz).
3. **Row-Factory:**  
   Nutzung von `sqlite3.Row` für eleganten, spaltenbasierten Zugriff (`row["topic_title"]` statt numerischer Indizes).
4. **Transaktions-Kapselung:**  
   Strikte Nutzung des Context-Managers (`with conn: ...`), sodass Schreiboperationen atomar comittet oder bei Fehlern vollständig zurückgerollt werden.

### 6.2 Mikroschritt: Interrogation der Library-Tabellen & Mandanten-Hierarchie

#### Das Pyramidenkonzept der Knowledge-Datenbank:
Die Wissensbasis gliedert sich in spezialisierte Register, die nach einem hierarchischen Prinzip abgefragt werden:

```text
                           ┌───────────────────────────┐
                           │   agent_system_manifest   │ ◄── Fundamental-Direktiven
                           └─────────────┬─────────────┘
                                         ▼
                           ┌───────────────────────────┐
                           │  meta_admin_taboo_rules   │ ◄── Sicherheits-Schranken
                           └─────────────┬─────────────┘
                                         ▼
                           ┌───────────────────────────┐
                           │     library_registry      │ ◄── Kern-Fähigkeiten (Pyramide)
                           └─────────────┬─────────────┘
                                         │
                 ┌───────────────────────┼───────────────────────┐
                 ▼                       ▼                       ▼
   ┌──────────────────────────┐ ┌──────────────────┐ ┌────────────────────────┐
   │ standard_column_library  │ │ def_library_reg  │ │ toolkit_library_reg    │
   │ Standard-Spaltenschemata │ │ Python-Funktionen│ │ Externe/Interne Tools  │
   └──────────────────────────┘ └──────────────────┘ └────────────────────────┘
```

#### Spezifikation der gezielten Abfrage-Logik:
1. **`library_registry`:**  
   Abfrage aktiver Kern-Fähigkeiten (`WHERE is_active = 1`). Extrahiert werden `topic_title`, `topic_specification`, `agent_instruction` und `example_code_snippet`.  
   *Zweck:* Liefert dem Agenten standardisierte Musterlösungen und Systemkonventionen.
2. **`standard_column_library_registry`:**  
   Abfrage der verbindlichen 19 Standard-Spalten (`node_id`, `is_active`, `year`...`target_node_id`).  
   *Zweck:* Garantiert, dass jede vom Agenten generierte Tabelle oder jeder neue Knoten zu 100% schemakonform ist.
3. **`def_library_registry`:**  
   Abfrage wiederverwendbarer Python-Funktionen.  
   *Zweck:* Verhindert das Neuerfinden des Rades; wenn bereits eine Hilfsfunktion für Dateiparser, SQLite-Inspektoren oder Hash-Checks existiert, wird diese als Code-Kontext geladen.
4. **`toolkit_library_registry` & `kernel_tools`:**  
   Abfrage registrierter Systemwerkzeuge und CLI-Befehle.  
   *Zweck:* Ermöglicht dem Agenten zu wissen, welche Module (`psutil`, `sqlite3`, `ollama`, `llama_cpp`) physisch verfügbar sind.

### 6.3 Mikroschritt: Validierung gegen System-Manifest & Taboo-Regeln

#### Schutz vor Halluzinationen und destruktiven Operationen:
Vor jeder Modellarbeit muss sichergestellt werden, dass die geplante Operation nicht gegen die Grundsätze des Systems verstößt.

#### 1. Direktiven aus `agent_system_manifest`:
* **`immutable_base_rule`:** Basis-Tabellen ohne dynamischen Zeitstempel dürfen niemals gelöscht (`DROP TABLE`) oder unkontrolliert überschrieben (`TRUNCATE`) werden.
* **`strict_schematic_binding`:** Verbot von dynamisch erfundenen Tabellennamen außerhalb der deklarierten Schemata. Alle neuen Tabellen müssen sich an das strikte Spaltenschema halten.
* **`closed_loop_verification`:** Kein generierter Programmcode darf ungeprüft an den Benutzer zurückgegeben werden. Jedes Code-Fragment muss vorab eine Syntax- und Sandbox-Prüfung durchlaufen.

#### 2. Restriktionen aus `meta_admin_taboo_rules`:
* **Taboo-Katalog:**  
  * Verbot von destruktiven Dateisystem-Befehlen (z. B. `rmdir /s`, `rm -rf`, unautorisiertes Löschen im Root-Verzeichnis).  
  * Verbot von ungesicherten Netzwerkverbindungen im Offline-Betrieb.  
  * Verbot von unendlichen Schleifen oder ungepufferten Speicherallokationen.
* **Injektion in den Denkprozess:**  
  Die Taboo-Regeln werden als negative Constraints formuliert und als fester Bestandteil in den Spezifikations-Kontext von JEV / LAYA eingespeist.

### 6.4 & 6.5 Mikroschritt: Lokaler Projekt-Scanner & Ähnlichkeits-Matching

#### 6.4 Lokaler Projekt-Kontext-Scanner (`Offline_AI`)
Neben der SQL-Datenbank besitzt das Projekt ein reichhaltiges Code-Ökosystem im Dateisystem:
1. **Traversierungs-Engine:**  
   Der Scanner inspiziert das Arbeitsverzeichnis `src/` (insbesondere `src/agent/agent_universal_skript/`, `src/docker_py/`, `src/Install/`).
2. **Klassen- & Methoden-Indexierung:**  
   * Schnelles Parsing von Klassennamen (z. B. `class_sql_reader`, `coreInfrastructure`, `OllamaManager`, `SubAgentVerifier`).  
   * Erfassung konfigurierter Konstanten (`DB_FILENAME`, `MAX_SANDBOX_TIMEOUT`, etc.).
3. **Vermeidung von Doppelarbeit:**  
   Verlangt die Nutzeranfrage beispielsweise das *"Lesen einer SQL Tabelle"*, erkennt der Scanner, dass `class_sql_reader.py` bereits robuste Methoden dafür bereitstellt, und weist das Modell an, diese bestehende Infrastruktur zu importieren und zu nutzen.

#### 6.5 Algorithmisches Ähnlichkeits-Matching (Relevanz-Filter)
Um das Modell nicht mit Hunderten irrelevanten Tabelleneinträgen zu überfluten, wird ein präziser Matching-Algorithmus spezifiziert:
* **TF-IDF & Keyword-Scoring:**  
  Die geschärfte Nutzeranfrage wird gegen die Spalten `topic_title`, `topic_specification` und `agent_instruction` aller Datensätze der Wissens-Tabellen abgeglichen.
* **Top-K Selektion:**  
  Nur die Top 3 bis 5 relevantesten Knoten mit einem Relevanz-Score über dem Schwellenwert (`TFIDF_SIMILARITY_THRESHOLD >= 0.15`) werden in den Spezifikations-Prompt aufgenommen.
* **Dynamische Kontext-Reduktion:**  
  Code-Beispiele (`example_code_snippet`) werden auf maximal 150 bis 200 Zeichen gekürzt, um das Token-Budget für JEV / LAYA optimal zu schonen.

### 6.6 Mikroschritt: Externe Wissensquellen & Web-Link Evaluierung

#### Parser für Web-Links und Dokumentationen:
* **Link-Erkennung:** Findet JEV/LAYA oder der Vorverarbeiter einen HTTP/HTTPS-Link in der Benutzeranfrage, wird ein automatischer Web-Scraping-Block (via `BeautifulSoup` oder `playwright`) aktiviert.
* **Inhalts-Evaluierung:** Der Agent liest die Website (z.B. eine API-Dokumentation oder ein StackOverflow-Thread), bewertet die Relevanz des Inhalts und fasst ihn zusammen.
* **Prompt-Veredelung:** Das extrahierte externe Wissen wird als `EXTERNAL_KNOWLEDGE` Abschnitt in den Prompt injiziert. Die ursprüngliche Nutzer-Anfrage wird dadurch erheblich aufgewertet und Fehler durch veraltetes LLM-Wissen werden verhindert.

## 🧠 Kapitel 9: Autonomes Lernen, Tool-Planung & Persistenz im Knowledge Tree

### 9.1 Idempotenz-Prüfung & Query-Cache in `user_query`
Ein lernfähiger Agent darf bekannte Aufgaben nicht jedes Mal rechenintensiv neu berechnen, sondern muss auf bestehende Erfahrungen zurückgreifen.

#### Cache-Prüfungs-Algorithmus:
1. **Hash- & Ähnlichkeits-Prüfung:**  
   Vor dem Aufruf von JEV/LAYA oder llama_cpp wird die geschärfte Abfrage mit bestehenden Datensätzen in `user_query` abgeglichen:  
   `SELECT node_id, agent_instruction, example_code_snippet, success_weight FROM user_query WHERE is_active = 1 AND topic_title LIKE ?`.
2. **Cache-Hit Kriterien:**  
   * Hohe textuelle Übereinstimmung (> 90%).  
   * `validation_rule = 'COMPLETED_CACHE_READY'` oder `'POSITIVE_VERIFIED'`.  
   * Positives Erfolgsgewicht (`success_weight >= 1.0`).
3. **Verhalten bei Cache-Hit:**  
   * Überspringen der Inferenz; sofortige Bereitstellung der bewährten Antwort im UI-Chatverlauf.  
   * Inkrementierung von `execution_count = execution_count + 1`.  
   * Enorme Einsparung von Rechenzeit und Hardware-Ressourcen.

### 9.2 Mikroschritt: Autonome Tool-Planung (`kernel_tools` & `toolkit_library_registry`)

#### Wenn die Aufgabe ein Werkzeug erfordert, das noch nicht existiert:
Stellt der Agent während der Kognitions- oder Ausarbeitungsphase fest, dass eine benötigte Operation (z. B. ein spezifischer CSV-Formatkonverter, ein API-Scanner oder ein mathematischer Prüfalgorithmus) fehlt, greift die **autonome Tool-Planung**:

```text
│    - Eindeutiger Tool-Name (z. B. "tool_csv_delimiter_detector")            │
│    - Funktions-Signatur mit strikten Typ-Annotationen                       │
│    - Erforderliche Python-Kernel-Module (z. B. "csv", "io")                 │
│    - Sicherheits- und Taboo-Einstufung                                      │
│                                                                             │
│ 3. CODE-SYNTHESE & MINIMAL-TEST:                                            │
│    - Erstellung einer modularen def-Funktion                                │
│    - Erzeugung eines synthetischen Unit-Tests                               │
│    - Sandbox-Ausführung via SubAgentVerifier                                │
│                                                                             │
│ 4. PERSISTENTE REGISTRIERUNG IN SQLITE:                                     │
│    - Eintrag in 'kernel_tools' / 'kernel_tool' mit Status 'ACTIVE'          │
│    - Verlinkung in 'toolkit_library_registry' (Node_ID, Ziel-Tabelle)        │
│    - Sofortige Verfügbarkeit für alle nachfolgenden Benutzeranfragen!       │
└─────────────────────────────────────────────────────────────────────────────┘
```

#### Schematische Verankerung des neuen Werkzeugs:
Das neue Tool wird mit allen 19 Standard-Spalten in `toolkit_library_registry` eingetragen:
* `node_id`: `TOOL_YYYYMMDD_HHMMSS`
* `topic_title`: Sprechender Name des Werkzeugs
* `topic_specification`: Detaillierte Funktions- und Parameter-Beschreibung
* `agent_instruction`: Konkrete Handlungsanweisung, wann und wie das Tool aufgerufen wird
* `example_code_snippet`: Lauffähiger Python-Aufrufcode
* `validation_rule`: `TOOL_SANDBOX_VERIFIED == TRUE`

### 9.3 & 9.4 Mikroschritt: Schematische Versiegelung in `user_query` & Audit-Protokollierung

#### 9.3 Strukturierte Speicherung in `user_query`
Jede neue, erfolgreich bearbeitete Anfrage wird dauerhaft in `user_query` verewigt und erscheint damit automatisch in der linken Seitenleiste unter *Recent Chats*:

```sql
INSERT INTO user_query (
    node_id, is_active, error_fallback_count,
    year, month, day, hour, minute, second,
    topic_title, topic_specification, agent_instruction,
    example_code_snippet, validation_rule, execution_count,
    success_weight, target_table, target_column_id, target_node_id
) VALUES (
    :node_id, 1, 0,
    :year, :month, :day, :hour, :minute, :second,
    :topic_title, :topic_specification, :agent_instruction,
    :example_code_snippet, 'COMPLETED_CACHE_READY', 1,
    1.0, 'library_registry', 'node_id', '000'
);
```

* **Vektorielle Zeitstempel:** Die Zerlegung des Datums in separate Integer-Spalten (`year`, `month`, `day`, `hour`, `minute`, `second`) ermöglicht blitzschnelle Zeitbereichs-Filterungen ohne rechenintensive Datums-Konvertierungen.
* **Inhalts-Trennung:**  
  * `topic_title`: Kurzer, prägnanter Fragetitel (max. 35 Zeichen).  
  * `topic_specification`: Der vom JEV/LAYA geschärfte und veredelte Prompt.  
  * `agent_instruction`: Die finale Antwort und Lösungsstrategie.  
  * `example_code_snippet`: Der verifizierte Programmcode.

#### 9.4 Deterministische Protokollierung in Log-Tabellen
Gemäß Systemstandard wird jeder Durchlauf zweifach versiegelt:
1. **Globales Master-Protokoll (`table_issue_data_time_stamp_run_protocol_for_query`):**  
   Chronologische Erfassung aller Vorgänge mit Laufzeit, Token-Verbrauch und Validierungsstatus.
2. **Dynamische Session-Log-Tabelle (`log_YYYYMMDD_HHMMSS`):**  
   Automatische Erstellung einer prozessspezifischen Log-Tabelle für den aktuellen Zeitstempel, in der Teilschritte (`init`, `enrich`, `execute`, `verify`, `seal`) zeilenweise protokolliert werden und live im rechten Virtual Studio Panel aufscheinen.

### 9.5 & 9.6 Mikroschritt: Dynamic Capability Promotion & Kontinuierliche Feedback-Schleife

#### 9.5 Promotion bewährter Lösungen zu neuen Fähigkeiten (`library_registry` & `def_library_registry`)
Nicht jede Abfrage ist nur eine flüchtige Frage – viele generierte Lösungen stellen wertvolles neues Systemwissen dar:
* **Beförderungs-Trigger:**  
  Wenn eine Lösung positiv validiert wurde und entweder vom Benutzer über den Button *[⚡ Als neue Fähigkeit verewigen]* markiert wird oder durch mehrfache erfolgreiche Ausführung auffällt (`execution_count >= 3`, `success_weight >= 1.4`).
* **Verankerung:**  
  Der Code wird als wiederverwendbarer Algorithmus in `def_library_registry` und als System-Fähigkeit in `library_registry` registriert:  
  `node_id = 'CAP_' || strftime('%Y%m%d_%H%M%S', 'now')`.  
  Damit steht die Lösung zukünftig allen Agenten im gesamten Projekt als natives Standardwissen zur Verfügung!

#### 9.6 Die Verstärkungs- und Rerouting-Schleife (Reinforcement from Feedback)
Das System optimiert sein Verhalten über ein dynamisches Belohnungssystem:

```text
       Benutzer-Feedback im ipywidgets Dashboard
                         │
        ┌────────────────┴────────────────┐
        ▼                                 ▼
   [👍 Daumen hoch]                 [👎 Daumen runter]
        │                                 │
        ▼                                 ▼
 - success_weight = +0.2           - success_weight = -0.5
 - validation_rule =               - Bei Gewicht <= 0:
   'POSITIVE_VERIFIED'               Knoten deaktiveren (is_active = 0)
 - Priorität bei zukünftigen       - Fehlerursache analysieren
   Suchabfragen steigt             - Eintrag in 'meta_admin_repair_registry'
                                   - Erstellung einer FX_CHAIN Reparatur-Route
```

* **Selbstheilung bei Fehlern:**  
  Ein negativ bewerteter Knoten wird nicht einfach gelöscht, sondern dient als negatives Trainingsbeispiel. Die Fehler-Signatur wird in `meta_admin_repair_registry` eingetragen, damit JEV / LAYA denselben Fehler bei ähnlichen Anfragen proaktiv vermeidet.

### 9.7 Mikroschritt: Integration industrieller Python-Werkzeuge (PandasAI, Sweetviz)

#### Dynamische Logik-Blöcke abhängig von der Frageform:
Die Architektur sieht vor, dass die *Form* der Benutzeranfrage alternative Analyse-Verhalten auslöst, bei denen mächtige Industrie-Tools eingeschleift werden.
* **Data Science Trigger:** Fragt der Nutzer 'Analysiere diese CSV-Daten', triggert das System nicht nur klassischen Code, sondern bindet **PandasAI** ein, um die Frage direkt als Pandas-Abfrage zu verarbeiten.
* **Visualisierungs & Profiling Trigger:** Bei Anfragen wie 'Wie sieht die Verteilung aus?', wird automatisch **Sweetviz** oder `ydata-profiling` aufgerufen, um tiefgreifende EDA-Reports (Exploratory Data Analysis) als HTML-Artefakte zu generieren und im UI (Tab *Files*) anzuzeigen.
* **Autonomes Tool-Routing:** JEV / LAYA entscheidet basierend auf der Intent-Klassifikation, ob ein solches High-Level Tool anstelle einer manuellen Code-Generierung verwendet werden soll.
